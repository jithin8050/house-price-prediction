```python
%%writefile app.py
import streamlit as st
import yfinance as yf
import pandas as pd

st.title("📈 Stock Portfolio Management System")

if "portfolio" not in st.session_state:
    st.session_state.portfolio = []

st.header("Add Stock")

stock_symbol = st.text_input("Stock Symbol (Example: TCS.NS)")
quantity = st.number_input("Quantity", min_value=1)
buy_price = st.number_input("Buy Price")

if st.button("Add Stock"):
    st.session_state.portfolio.append({
        "Stock": stock_symbol,
        "Quantity": quantity,
        "Buy Price": buy_price
    })
    st.success("Stock Added Successfully")

st.header("Portfolio")

if len(st.session_state.portfolio) > 0:

    df = pd.DataFrame(st.session_state.portfolio)

    current_prices = []
    profits = []

    for i, row in df.iterrows():

        ticker = yf.Ticker(row["Stock"])
        data = ticker.history(period="1d")

        if len(data) > 0:
            current_price = data["Close"].iloc[-1]
        else:
            current_price = 0

        current_prices.append(current_price)

        profit = (current_price - row["Buy Price"]) * row["Quantity"]
        profits.append(profit)

    df["Current Price"] = current_prices
    df["Profit/Loss"] = profits

    st.dataframe(df)

    total_profit = sum(profits)

    if total_profit > 0:
        st.success(f"Total Profit: ₹{round(total_profit,2)}")
    else:
        st.error(f"Total Loss: ₹{round(total_profit,2)}")

else:
    st.write("Portfolio Empty")
```
