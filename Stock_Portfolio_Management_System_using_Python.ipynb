```python
import yfinance as yf
import pandas as pd

portfolio = []

def add_stock():
    stock = input("Enter stock symbol (Example: TCS.NS): ")
    quantity = int(input("Enter quantity: "))
    buy_price = float(input("Enter buy price: "))

    portfolio.append({
        "Stock": stock,
        "Quantity": quantity,
        "Buy Price": buy_price
    })

def show_portfolio():
    if len(portfolio) == 0:
        print("Portfolio is empty")
        return
    df = pd.DataFrame(portfolio)
    print(df)

def check_profit_loss():
    total_investment = 0
    total_current = 0

    for stock in portfolio:

        ticker = yf.Ticker(stock["Stock"])
        current_price = ticker.info["regularMarketPrice"]

        buy_value = stock["Buy Price"] * stock["Quantity"]
        current_value = current_price * stock["Quantity"]

        profit_loss = current_value - buy_value

        print("\nStock:", stock["Stock"])
        print("Current Price:", current_price)
        print("Quantity:", stock["Quantity"])

        if profit_loss > 0:
            print("Profit:", round(profit_loss,2))
        else:
            print("Loss:", round(abs(profit_loss),2))

        total_investment += buy_value
        total_current += current_value

    total_profit = total_current - total_investment

    print("\nTotal Investment:", total_investment)
    print("Current Value:", total_current)

    if total_profit > 0:
        print("Overall Profit:", round(total_profit,2))
    else:
        print("Overall Loss:", round(abs(total_profit),2))


while True:

    print("\n1 Add Stock")
    print("2 View Portfolio")
    print("3 Check Profit/Loss (Real Time)")
    print("4 Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        add_stock()

    elif choice == "2":
        show_portfolio()

    elif choice == "3":
        check_profit_loss()

    elif choice == "4":
        break
```
